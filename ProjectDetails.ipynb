### SPSS Transformation API

I built an SPSS Variable Renamer and Metadata Transformation API using FastAPI and MongoDB.

The API allowed users to upload SPSS .sav files and CSV files containing transformation rules. It automatically renamed variables, updated metadata, and exported the final data into Excel or CSV files while keeping the original labels and formats.

I used MongoDB to store upload details and validation errors, which helped with tracking and auditing.

The main challenge was keeping the SPSS metadata correct while renaming variables. I solved this by carefully mapping the variable labels and formats during the transformation.

Overall, the project reduced manual work, saved time, and improved data quality for analysts.

### AutoVerb Tool

I built AutoVerb to automate code frame generation for survey responses.

### RAG Project

In my RAG project, I worked with large PDF files and divided them into 
smaller meaningful chunks . I created embeddings for these chunks and
stored them in ChromaDB. When a user asks a question, the system searches for the
 most relevant sections and gives them to the LLM to generate the answer.

I also used Agentic AI with LangChain to make the system smarter.
The agents could break complex questions into smaller steps, 
check the retrieved information, and provide the right context to the LLM.
This made the system more automatic, reliable, and easier to scale.

## 2. Project details(FastAPI, MongoDB, SPSS Transformation)

#### Situation:  

Our survey team worked with SPSS .sav files, which contain survey data in complex metadata (like variable labels, formats, missing values). Renaming variables and transforming metadata manually thats time‑consuming.


#### Task: 

I was design API that automatically rename variables, apply transformation rules, and preserve all metadata and also validating inputs to prevent errors.


#### Action :

Built the API using FastAPI for high‑performance endpoints.

Integrated MongoDB to log file uploads, validation errors, and transformation history for auditing and debugging.

Implemented validation layers to catch issues like missing columns, duplicate variable names, invalid references, and incorrect query formats.

Designed endpoints to process .sav files, apply transformation rules from CSV templates, and export results in Excel/CSV formats.

Used streaming responses for large file downloads to improve performance.

Deployed the application with Docker and AWS for scalability.


#### Challenges Faced :

Preserving metadata: SPSS files store labels, formats, and missing values — renaming variables without losing this was tricky.

Performance: Large .sav files required memory optimization and streaming to avoid timeouts.
