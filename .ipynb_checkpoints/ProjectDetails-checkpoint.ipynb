## 1. Project details (OE,RAG,LangChain,LLM,ChromaDB)

Built an OpenEnd Database system using FastAPI for the backend.

Processed a Large PDF and split it into semantic chunks.

Generated unique embeddings for each chunk using OpenAI embeddings.

Stored embeddings in ChromaDB (Vector Database) for efficient similarity search.

Integrated LangChain to orchestrate the retrieval process and connect with LLMs.

Implemented RAG (Retrieval‑Augmented Generation) so that user queries could retrieve the closest chunks and provide accurate, context‑aware answers.

Containerized the application with Docker and deployed it on AWS Linux for scalability.


#### Challenges Faced

Embedding size and performance: Large embeddings slowed down queries → solved with indexing and caching.

## 2. Project details(FastAPI,MongoDB,SPSS Transformation)

#### Situation:  

Our survey team often worked with SPSS .sav files, which contain survey data along with complex metadata (like variable labels, formats, missing values). Renaming variables or transforming metadata manually was error‑prone and time‑consuming.


#### Task: 

I was responsible for designing an API that could automatically rename variables, apply transformation rules, and preserve all metadata — while also validating inputs to prevent errors.


#### Action :

Built the API using FastAPI for high‑performance endpoints.

Integrated MongoDB to log file uploads, validation errors, and transformation history for auditing and debugging.

Implemented validation layers to catch issues like missing columns, duplicate variable names, invalid references, and incorrect query formats.

Designed endpoints to process .sav files, apply transformation rules from CSV templates, and export results in Excel/CSV formats.

Used streaming responses for large file downloads to improve performance.

Deployed the application with Docker and AWS for scalability.


#### Challenges Faced :

Preserving metadata: SPSS files store labels, formats, and missing values — renaming variables without losing this was tricky.

Performance: Large .sav files required memory optimization and streaming to avoid timeouts.
